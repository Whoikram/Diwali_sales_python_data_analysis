```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
df=pd.read_csv('Diwali Sales Data.csv',encoding='unicode_escape')
```

```python
df.shape
```

```python
df.head()
```

```python
df.info()
```

```python
df.drop(['Status','unnamed1'],axis=1,inplace=True)
```

```python
df.info()
```

```python
pd.isnull(df).sum()
```

```python
df.dropna(inplace=True)
#drop null values
```

```python
pd.isnull(df).sum()
```

```python
#used to change data Type of Amount column fromm float to int
df['Amount']= df['Amount'].astype('int')
```

```python
df['Amount'].dtype
```

```python
df.columns
```

```python
#rename column
df.rename(columns={'Marital_Status':'Shadi'})
```

```python
df.describe()
```

```python
df[['Age','Orders']].describe()
#to count min max count etc
```

```python
ax= sns.countplot(x='Gender',data=df)
```

```python
df.groupby(['Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
```

```python
sales_gen= df.groupby(['Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.barplot(x='Gender', y='Amount',data=sales_gen)
```

```python
from the above graph we can see that most of the buyers are female and even purchasing power of females are greater than men
```

```python
df.columns
```

```python
ax=sns.countplot(x='Age Group',data=df,hue='Gender')
```

```python
sales_age= df.groupby(['Age Group'], as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.barplot(x= 'Age Group', y= 'Amount',data=sales_age)
```

```python
from the graph we can see that most of the buyrs are age group between 26-35 yrs female
```

```python
sales_state= df.groupby(['State'], as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(10)
sns.set(rc={'figure.figsize':(15,5)})
sns.barplot(x='State', y='Orders',data=sales_state)
```

```python
sales_state= df.groupby(['State'], as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False).head(10)
sns.set(rc={'figure.figsize':(15,5)})
sns.barplot(x='State', y='Amount',data=sales_state)
```

```python
from the above graphs we can see that most of the orders are from Uttarpradesh , Maharashtra and Karnataka respectively
```

```python
ax= sns.countplot(x='Marital_Status',data=df)
```

```python
marital = df.groupby(['Marital_Status','Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.set(rc={'figure.figsize':(6,5)})
sns.barplot(data= marital, x='Marital_Status', y='Amount',hue='Gender')
```

```python
sales_state= df.groupby(['Occupation'], as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(x='Occupation', y='Amount',data=sales_state)
```

```python
from the graph we can see that most of the buyers are working in IT ,Healthcare and Aviation respectively
```

```python
df.columns
```

```python
sales_state= df.groupby(['Product_Category'], as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.set(rc={'figure.figsize':(40,5)})
sns.barplot(x='Product_Category', y='Amount',data=sales_state)
```

```python
MOST OF THE SOLD PRODUCTS ARE FROM FOOD , CLOTHING AND ELECTRONICS category
```

```python
#top Selling Products
sales_state= df.groupby(['Product_ID'], as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(10)
sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(x='Product_ID', y='Orders',data=sales_state)
```

```python
CONCLUSION –
UNMARRIED WOMEN AGED 26–35 FROM UTTAR PRADESH, MAHARASHTRA, AND KARNATAKA, WORKING IN THE IT AND HEALTHCARE SECTORS, SPENT THE MOST DURING THIS DIWALI SALE.
FOOD, CLOTHING, AND ELECTRONIC GADGETS WERE THE MOST COMMON PURCHASES.
NEXT YEAR, OUR MAIN FOCUS WILL BE ON THESE CATEGORIES...

```
